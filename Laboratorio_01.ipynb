## 3. Funzioni e accumulatori

```julia
function fact(n)
    # calcola il fattoriale di n
    # n dev'essere un intero
    f = 1.0
    # f fa da accumulatore: parte da 1,
    # a ogni passo, lo moltiplico per k
    for k = 1:n
        f = f * k
    end
    # ora f vale n!
    return f
end

println("4! = ", fact(4))
```

In un notebook la funzione non va salvata in un file separato: dopo aver eseguito la cella che la definisce, possiamo richiamarla da qualunque cella successiva. L'inizializzazione con `1.0` fa sì che l'accumulatore sia un `Float64`, come i numeri usati nel programma MATLAB originale.

```julia
fact(10)
```

## 4. Calcolo dell'esponenziale

$x^n=1+\frac{x}{1!}+\dots+\frac{x^k}{k!}$

```julia
function myexp(x, n)
    # calcola exp(x) con la serie di Taylor troncata all'n-esimo termine
    a = 1.0  # accumulatore
    for k = 1:n
        a = a + pow(x, k) / fact(k)
    end
    return a
end
```

Qualche esperimento su quanti termini servono per approssimare bene: confrontiamo con la funzione `exp(x)` di Julia, che calcola l'esponenziale meglio di noi.

```julia
function myexp(x, n)
    # calcola exp(x) con la serie di Taylor troncata all'n-esimo termine
    a = 1.0  # accumulatore
    for k = 1:n
        a = a + pow(x, k) / fact(k)
    end
    return a
end

@show myexp(1.0, 5)
@show myexp(10.0, 50)
@show exp(10.0);
@show myexp(-20.0, 500);
```

Due problemi:

- **Inaccurato:** provate `myexp(-20.0, 500)`.
- **Lento:** con $n$ termini, il numero di operazioni da eseguire cresce come $n^2$ (perché?).

Risolviamo il secondo problema introducendo un altro accumulatore:

anziche fare $a =a+\frac{x^k}{k!}$ creo $\frac{x^k}{k!}$ facendo $t=t\cdot\frac{x}{k}$ con $k$ che cresce e poi $a=a+t$ per creare la somma di taylor

```julia
function myexp2(x, n)
    # calcola e^x con Taylor troncato
    # ma usa solo O(n) operazioni
    t = 1.0  # accumulatore che contiene il termine generico della sommatoria
    a = 1.0  # accumulatore che contiene le somme parziali
    for k = 1:n
        t = t * x / k
        a = a + t
    end
    return a
end

@show myexp(-3,500);
@show myexp(3,500);
```

Ora va meglio:

```julia
@show myexp(-20.0, 500)
@show myexp2(-20.0, 500);
```

Cosa succedeva?

```julia
@show fact(500)
@show pow(-20.0, 500)
@show Inf / Inf;
```

Ci sono ancora problemi di accuratezza sui numeri negativi:

```julia
myexp2(-30.0, 500)
```

Un esponenziale dovrebbe sempre essere positivo… Ci sono pesanti errori di cancellazione nella formula che abbiamo usato (perché?).

$$
\text{Errore relativo:}\quad \frac{\left|\operatorname{myexp2}(x)-e^x\right|}{e^x}.
$$

La soluzione: cambiare algoritmo e sceglierne uno che non porti a errori di cancellazione.

```julia
@show exp(-30.0)
@show myexp2(-30.0, 500)
@show 1 / myexp2(30.0, 500);
```

### Esercizio 3

Scrivere una funzione `solve2(a, b, c)` che risolva l'equazione di secondo grado $ax^2+bx+c=0$ nel modo più stabile possibile. Testare l'implementazione su $x^2-(10^{10}+10^{-10})x+1=0$.

**Suggerimento:** ci sono al massimo due sottrazioni. Una è necessaria (perché?); l'altra no.

```julia
# Scrivete qui la funzione solve2(a, b, c) e il test richiesto

function solve2(a, b, c)
  delta = sqrt((b^2)-(4*a*c))
  # evito la cancellazione numerica usando sign(b) per avere sempre -b-delta o b+delta
  x2 = (-b-sign(b)*delta)/(2*a)
  x1=c/(a*x2)

return x1,x2
end

@show solve2(1,-1,-6);
@show solve2(1,-2,1);
```
